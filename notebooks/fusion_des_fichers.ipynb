```python
import pandas as pd

# ==========================
# Charger les datasets
# ==========================
players = pd.read_csv("sofascore_database_globale.csv")
stats = pd.read_csv("sofascore_top5leagues_players_stat.csv")

# ==========================
# Nettoyer les noms
# ==========================
players["name_clean"] = (
    players["name"]
    .astype(str)
    .str.lower()
    .str.strip()
)

stats["name_clean"] = (
    stats["Nom"]
    .astype(str)
    .str.lower()
    .str.strip()
)

# ==========================
# Fusionner les datasets
# ==========================
merged = pd.merge(
    players,
    stats,
    on="name_clean",
    how="inner"      # "left" si tu veux garder tous les joueurs
)

# ==========================
# Supprimer les colonnes répétitives
# ==========================
colonnes_a_supprimer = [
    "name_clean",
    "Nom",
    "Ligue",
    "Equipe"
]

merged.drop(columns=colonnes_a_supprimer, inplace=True, errors="ignore")

# ==========================
# Réorganiser les colonnes
# ==========================
ordre = [
    "player_id",
    "name",
    "position",
    "age",
    "country",
    "Buts",
    "xG",
    "Succ_dribbles",
    "Tacles",
    "Assists",
    "Passes_Reussies_Pct",
    "club",
    "league",
    "market_value_raw",
]

# Garder seulement les colonnes existantes
ordre = [c for c in ordre if c in merged.columns]
merged = merged[ordre]

# ==========================
# Sauvegarder
# ==========================
merged.to_csv("dataset_final.csv", index=False, encoding="utf-8-sig")

print("Fusion terminée.")
print("Nombre de joueurs :", len(merged))
```

```python
import ipywidgets as widgets
from IPython.display import display
import pandas as pd
import joblib

# Charger le modèle
model = joblib.load("modele_marche_joueur.pkl")

# Création des widgets de saisie
pos_widget = widgets.Dropdown(options=['F', 'M', 'D','G'], value='F', description='Poste:')
age_widget = widgets.IntSlider(value=24, min=16, max=40, description='Âge:')
buts_widget = widgets.IntText(value=10, description='Buts:')
xg_widget = widgets.FloatText(value=9.5, description='xG:')
dribbles_widget = widgets.IntText(value=30, description='Dribbles:')
tacles_widget = widgets.IntText(value=15, description='Tacles:')
assists_widget = widgets.IntText(value=5, description='Assists:')
passes_widget = widgets.FloatSlider(value=82.0, min=0, max=100, description='Passes %:')
country_widget = widgets.Text(value='France', description='Pays:')
club_widget = widgets.Text(value='bayern_munich', description='Club:')
league_widget = widgets.Text(value='bundesliga', description='Ligue:')

button = widgets.Button(description="Calculer la Valeur", button_style='success')
output = widgets.Output()

def on_button_clicked(b):
    with output:
        output.clear_output()
        nouveau_joueur = pd.DataFrame([{
            "position": pos_widget.value,
            "age": float(age_widget.value),
            "country": country_widget.value,
            "Buts": int(buts_widget.value),
            "xG": float(xg_widget.value),
            "Succ_dribbles": int(dribbles_widget.value),
            "Tacles": int(tacles_widget.value),
            "Assists": int(assists_widget.value),
            "Passes_Reussies_Pct": float(passes_widget.value),
            "club": club_widget.value,
            "league": league_widget.value
        }])
        valeur_predite = model.predict(nouveau_joueur)[0]
        print(f" Valeur marchande estimée : {valeur_predite:,.2f} €")

button.on_click(on_button_clicked)

# Affichage de l'interface dans Jupyter
display(pos_widget, age_widget, buts_widget, xg_widget, dribbles_widget, 
        tacles_widget, assists_widget, passes_widget, country_widget, 
        club_widget, league_widget, button, output)
```
